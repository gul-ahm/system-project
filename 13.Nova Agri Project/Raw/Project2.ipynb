```python
import pandas as pd

# Load the sales register CSV file
sales = pd.read_csv('D:\Insightrix Project\Finacial dataset\sales_register.csv')

# Load the receipts register CSV file
receipt = pd.read_csv('D:\Insightrix Project\Finacial dataset\Reciepts_register.csv')

# Display the first few rows of each DataFrame
print("Sales Register:")
print(sales.head())

print("\nReceipts Register:")
print(receipt.head())
```

```python
import plotly.express as px

# Count the occurrences of each Particulars_id in the sales and receipts datasets
sales_counts = sales['Particulars_id'].value_counts().reset_index()
sales_counts.columns = ['Particulars_id', 'Count']

receipts_counts = receipt['Particulars_id'].value_counts().reset_index()
receipts_counts.columns = ['Particulars_id', 'Count']

# Plot the counts using plotly
fig_sales = px.bar(sales_counts, x='Particulars_id', y='Count', title='Sales Particulars_id Counts')
fig_receipts = px.bar(receipts_counts, x='Particulars_id', y='Count', title='Receipts Particulars_id Counts')

# Combine the sales and receipts counts into a single DataFrame for comparison
combined_counts = pd.merge(sales_counts, receipts_counts, on='Particulars_id', suffixes=('_Sales', '_Receipts'))

# Plot the combined counts using plotly
fig_combined = px.bar(combined_counts, x='Particulars_id', y=['Count_Sales', 'Count_Receipts'], 
                      title='Comparison of Sales and Receipts Particulars_id Counts')

fig_combined.show()
fig_receipts.show()

# Save the counts to CSV files
sales_counts.to_csv('D:\\Insightrix Project\\Pr2\\sales_particulars_counts.csv', index=False)
receipts_counts.to_csv('D:\\Insightrix Project\\Pr2\\receipts_particulars_counts.csv', index=False)
```

```python
# Group the sales and receipt data by Particulars_id and sum the values
sales_total = sales.groupby('Particulars_id')['Value'].sum().reset_index()
sales_total.columns = ['Particulars_id', 'Total_Sales']

receipts_total = receipt.groupby('Particulars_id')['Credit'].sum().reset_index()
receipts_total.columns = ['Particulars_id', 'Total_Receipts']

# Merge the grouped data into a single DataFrame
total_amounts = pd.merge(sales_total, receipts_total, on='Particulars_id', how='outer').fillna(0)

# Save the merged DataFrame to a CSV file
total_amounts.to_csv('D:\\Insightrix Project\\Pr2\\total_amounts_by_particulars_id.csv', index=False)
```

```python
# Extract the date and Particulars_id columns from the sales and receipt data
sales_dates = sales[['Particulars_id', 'Date']]
sales_dates.columns = ['Particulars_id', 'Date from Sales Register']

receipt_dates = receipt[['Particulars_id', 'Date']]
receipt_dates.columns = ['Particulars_id', 'Date from Receipt Register']

# Merge the two DataFrames on Particulars_id
merged_dates = pd.merge(sales_dates, receipt_dates, on='Particulars_id', how='outer')

# Save the merged DataFrame to a CSV file
merged_dates.to_csv('D:\\Insightrix Project\\Pr2\\merged_dates_by_particulars_id.csv', index=False)
```

```python
# Extract unique Particulars_id and TAX_ID from the sales data
unique_ids = sales[['Particulars_id', 'TAX_ID']].dropna().drop_duplicates()

# Count the occurrences of each TAX_ID
tax_id_counts = unique_ids['TAX_ID'].value_counts().reset_index()
tax_id_counts.columns = ['TAX_ID', 'Count']

# Merge the counts with the unique_ids DataFrame to include Particulars_id
tax_id_counts = pd.merge(tax_id_counts, unique_ids, on='TAX_ID', how='left')

# Save the result to a CSV file
tax_id_counts.to_csv('D:\\Insightrix Project\\Pr2\\tax_id_counts_with_particulars_id.csv', index=False)

# Print the result
print(tax_id_counts)
```

```python
import pandas as pd
import folium
from geopy.geocoders import Nominatim
from time import sleep

# Load sales data
file_path = "D:\Insightrix Project\Finacial dataset\sales_register.csv"
df = pd.read_csv(file_path)

# Extract required columns
filter_columns = ["Particulars_id", "TAX_ID", "Location"]  # Update column names if different
df = df[filter_columns].dropna()

# Initialize geocoder
geolocator = Nominatim(user_agent="geo_mapper")

# Function to get latitude and longitude
def get_lat_lon(location):
    try:
        loc = geolocator.geocode(location, timeout=10)
        if loc:
            return loc.latitude, loc.longitude
    except Exception as e:
        print(f"Error geocoding {location}: {e}")
    return None, None

# Apply geocoding and add latitude, longitude columns
df["Latitude"], df["Longitude"] = zip(*df["Location"].apply(get_lat_lon))
df.dropna(subset=["Latitude", "Longitude"], inplace=True)  # Remove failed geocodes

# Save geocoded data (to avoid redundant API calls)
df.to_csv("D:\Insightrix Project\Pr2", index=False)

# Create an interactive Folium map
m = folium.Map(location=[df["Latitude"].mean(), df["Longitude"].mean()], zoom_start=5)

# Add markers
for _, row in df.iterrows():
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=f"ID: {row['Particulars_id']}<br>TAX: {row['TAX_ID']}<br>Location: {row['Location']}",
    ).add_to(m)

# Save the map
m.save("D:\Insightrix Project\Pr2")
print("Map saved! Open 'sales_map.html' to view it.")

```
