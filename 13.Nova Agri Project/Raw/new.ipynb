```python
import pandas as pd
sales = pd.read_csv('D:\Insightrix Project\Finacial dataset\sales_register.csv')
```

```python
# Group by 'Particulars_id' and 'ds' and count the occurrences
particulars_count = sales.groupby(['Particulars_id', 'ds']).size().reset_index(name='count')

# Save the result to a new CSV file
particulars_count.to_csv('particulars_count.csv', index=False)
```

```python
import pandas as pd

# Given data
data = {
    "Value": [
         94835, 17458.4
    ],
    "Gross Total": [
        111905.00, 20601.00
	


    ]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Calculate tax percentage
df["Tax Percentage"] = ((df["Gross Total"] - df["Value"]) / df["Value"]) * 100

# Display the result
df[["Value", "Gross Total", "Tax Percentage"]]

# Export the DataFrame to a CSV file
df.to_csv('calculated_tax_percentage.csv', index=False)
df[["Value", "Gross Total", "Tax Percentage"]]
```
